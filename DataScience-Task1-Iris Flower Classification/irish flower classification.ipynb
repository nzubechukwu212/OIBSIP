```python
#import necessary libraries
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
#load the iris dataset
iris = load_iris()
#display the features and target names
print("Features:", iris.feature_names)
print("Target Names:", iris.target_names)
display(iris.data, iris.target)   # Display the data and target values
```

```python
# perform EDA (Exploratory Data Analysis)
iris.data.shape

```

```python
#data types

print("Data Types:")
print("Iris Data:", iris.data.dtype)
print("Iris Target:", iris.target.dtype)
```

```python
#check for missing values
import numpy as np
print("Missing Values in Iris Data:", np.isnan(iris.data).sum())
```

```python
#descriptive statistics
import pandas as pd
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("Descriptive Statistics:")
print(df.describe())
```

```python
#data visualization
#pair plot showing feature
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target # add target as a column
sns.pairplot(df, hue='species', markers=["o", "s", "D"])
plt.show()
```

```python
#boxplot for each feature
for i, feature in enumerate(iris.feature_names):
    plt.figure(i)
    sns.boxplot(x='species', y=feature, data=df)
    plt.title(f'Boxplot of {feature} by Species')
    plt.show()
```

```python
#feature selection using correlation matrix
plt.figure(figsize=(10, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.show()
```

```python
#which features are most descriminative
discriminative_features = correlation_matrix.abs().idxmax()
print("Most Discriminative Features:")
print(discriminative_features)
```

```python
#define the features and target variable
X = iris.data
y = iris.target
```

```python
#train-test split
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
```

```python
#train a machine learning model
#traning a logistic regression model
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
#Evaluate logistic regression model
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:")
print(classification_report(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred)) 
```

```python
#display the confusion matrix as a heatmap

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, cmap="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
#training a random forest model
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

```python
#evaluate random forest model
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:")
print(classification_report(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
#visualize the confusion matrix for random forest model
plt.figure(figsize=(8, 6))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, cmap="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
#compare the performance of the logistic regression model and the random forest model
#comparing model performance
print("Model Comparison")
print("-" * 40)
print(f"{'Metric':<12}{'Logistic Regression':<22}{'Random Forest'}")
print(f"{'Accuracy':<12}{1.0:<22}{1.0}")
print(f"{'Precision':<12}{1.0:<22}{1.0}")
print(f"{'Recall':<12}{1.0:<22}{1.0}")
print(f"{'F1-Score':<12}{1.0:<22}{1.0}")
```

## Best Model

Both Logistic Regression and Random Forest achieved perfect scores (accuracy, 
precision, recall, F1 = 1.0) on the test set. This isn't surprising — Iris is 
a small (150 rows), clean, and well-separated dataset, especially for setosa, 
which is linearly separable from the other two species.

Chosen model: Logistic Regression

Since both models perform identically here, the tie-breaker is model 
simplicity, not raw metrics:
- Logistic Regression is simpler, faster to train, and easier to interpret 
  (its coefficients directly show feature influence).
- Random Forest is a more powerful, flexible model, but that flexibility is 
  unnecessary on a dataset this small and simple — it adds complexity without 
  adding accuracy.

On a harder or larger dataset, Random Forest would likely pull ahead, but 
for Iris specifically, the simpler model is the more justified choice.
